```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
```

```python
# Configuration
num_records = 100
start_time = datetime.now()
data_list = []

# Simulation Loop
for i in range(num_records):
    timestamp = start_time + timedelta(minutes=i*5) # Data every 5 minutes
    
    sensor_id = "SN-001"
    temperature = round(np.random.uniform(27.5, 33.0), 2) # Celsius
    humidity = round(np.random.uniform(70.0, 85.0), 2) # Percentage
    co2_level = int(np.random.normal(400, 50)) # PPM

    data_list.append({
        "timestamp": timestamp,
        "sensor_id": sensor_id,
        "temperature": temperature,
        "humidity": humidity,
        "co2_level": co2_level
    })

# Create DataFrame
df = pd.DataFrame(data_list)

# Preview the data
print(df.head())
```

```python
# Save to CSV
df.to_csv("environmental_data.csv", index=False)
df.to_json("environmental_data.json", orient="records")
print("Files saved successfully!")
```
